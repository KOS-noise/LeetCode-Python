<h2 style="color:#0969da;">28. Find the Index of the First Occurrence in a String</h2>

[LeetCode 문제 링크](https://leetcode.com/problems/find-the-index-of-the-first-occurrence-in-a-string/)

Given two strings <code>needle</code> and <code>haystack</code>, return the index of the first occurrence of <code>needle</code> in <code>haystack</code>, or <code>-1</code> if <code>needle</code> is not part of <code>haystack</code>.

---

<h3 style="color:#8250df;">Example 1</h3>

- **Input:** `haystack = "sadbutsad"`, `needle = "sad"`
- **Output:** `0`
- **Explanation:** `"sad"` occurs at index 0 and 6. The first occurrence is at index 0.

<h3 style="color:#8250df;">Example 2</h3>

- **Input:** `haystack = "leetcode"`, `needle = "leeto"`
- **Output:** `-1`
- **Explanation:** `"leeto"` did not occur in `"leetcode"`, so we return `-1`.

---

<h3 style="color:#1a7f37;">Constraints</h3>

- `1 <= haystack.length, needle.length <= 10^4`
- `haystack` and `needle` consist of only lowercase English characters.

```python
class Solution:
    def strStr(self, haystack: str, needle: str) -> int:
      
        if needle in haystack:
            p = haystack.find(needle)
            return p
        else:
            return -1




sol = Solution()
print(sol.strStr("sadbutsad", "sad"))
print(sol.strStr("leetcode", "leeto"))
```

그동안 풀었던 것 중에 제일 쉬웠다. 
find() 함수를 사용하면 어느 위치에서 해당 문자열이 나오는지 인덱스를 알 수 있다.
