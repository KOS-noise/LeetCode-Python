<h2 style="color:#0969da;">13. Roman to Integer</h2>

[LeetCode 문제 링크](https://leetcode.com/problems/roman-to-integer/)

Roman numerals are represented by seven different symbols: `I`, `V`, `X`, `L`, `C`, `D` and `M`.

| Symbol | Value |
|--------|-------|
| I | 1 |
| V | 5 |
| X | 10 |
| L | 50 |
| C | 100 |
| D | 500 |
| M | 1000 |

For example, 2 is written as `II` in Roman numeral, just two ones added together. 12 is written as `XII`, which is simply `X + II`. The number 27 is written as `XXVII`, which is `XX + V + II`.

Roman numerals are usually written largest to smallest from left to right. However, the numeral for four is not `IIII`. Instead, the number four is written as `IV`. Because the one is before the five we subtract it making four. The same principle applies to the number nine, which is written as `IX`. There are six instances where subtraction is used:

- `I` can be placed before `V` (5) and `X` (10) to make 4 and 9.
- `X` can be placed before `L` (50) and `C` (100) to make 40 and 90.
- `C` can be placed before `D` (500) and `M` (1000) to make 400 and 900.

Given a roman numeral, convert it to an integer.

---

<h3 style="color:#8250df;">Example 1</h3>

- **Input:** `s = "III"`
- **Output:** `3`
- **Explanation:** `III = 3`

<h3 style="color:#8250df;">Example 2</h3>

- **Input:** `s = "LVIII"`
- **Output:** `58`
- **Explanation:** `L = 50`, `V = 5`, `III = 3`

<h3 style="color:#8250df;">Example 3</h3>

- **Input:** `s = "MCMXCIV"`
- **Output:** `1994`
- **Explanation:** `M = 1000`, `CM = 900`, `XC = 90` and `IV = 4`

---

<h3 style="color:#1a7f37;">Constraints</h3>

- `1 <= s.length <= 15`
- `s` contains only the characters (`'I'`, `'V'`, `'X'`, `'L'`, `'C'`, `'D'`, `'M'`).
- It is guaranteed that `s` is a valid roman numeral in the range `[1, 3999]`.

```python
class Solution:
    def romanToInt(self, s: str) -> int:
        value = {
            "I": 1,
            "V": 5,
            "X": 10,
            "L": 50,
            "C": 100,
            "D": 500,
            "M": 1000,
        }

        total = 0
        for i in range(len(s)):
            if i + 1 < len(s) and value[s[i]] < value[s[i + 1]]:
                total -= value[s[i]]
            else:
                total += value[s[i]]
        return total


sol = Solution()

# 1) 리트코드처럼: 입력을 직접 넣어서 함수 호출
print(sol.romanToInt("III"))
print(sol.romanToInt("LVIII"))
print(sol.romanToInt("MCMXCIV"))

# 2) 콘솔에서 직접 입력받고 싶을 때 (리트코드에선 안 써도 됨)
# s = input("roman: ")
# print(sol.romanToInt(s))
```

<h2 style="color:#8250df;">해설 (Roman to Integer)</h2>

로마 숫자는 보통 왼쪽이 크고 오른쪽이 작다. 그때는 그냥 더하면 된다.

예외는 **작은 값이 큰 값 앞에 올 때**다. 이때는 빼야 한다.

- `VI` → 5 다음에 1. 5가 더 크니까 `5 + 1 = 6`
- `IV` → 1 다음에 5. 1이 더 작으니까 `5 - 1 = 4`

코드는 `s`를 인덱스 `i`로 한 글자씩 본다.

1. `value` 딕셔너리로 글자를 숫자로 바꾼다.
2. 다음 글자가 있고, **지금 값 < 다음 값**이면 지금 값을 뺀다.
3. 아니면 지금 값을 더한다.
4. 마지막 글자는 다음이 없으니 무조건 더한다.

`i + 1 < len(s)` 는 다음 글자가 있는지 확인하는 조건이다. 없으면 `s[i + 1]`을 볼 때 에러가 난다.

**`MCMXCIV`**

| i | 글자 | 값 | 다음 | 동작 | total |
|---|------|----|------|------|-------|
| 0 | M | 1000 | C 100 | 더하기 | 1000 |
| 1 | C | 100 | M 1000 | 빼기 | 900 |
| 2 | M | 1000 | X 10 | 더하기 | 1900 |
| 3 | X | 10 | C 100 | 빼기 | 1890 |
| 4 | C | 100 | I 1 | 더하기 | 1990 |
| 5 | I | 1 | V 5 | 빼기 | 1989 |
| 6 | V | 5 | 없음 | 더하기 | 1994 |
