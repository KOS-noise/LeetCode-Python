<h2 style="color:#0969da;">14. Longest Common Prefix</h2>

[LeetCode 문제 링크](https://leetcode.com/problems/longest-common-prefix/)

Write a function to find the longest common prefix string amongst an array of strings.

If there is no common prefix, return an empty string <code>""</code>.

---

<h3 style="color:#8250df;">Example 1</h3>

- **Input:** `strs = ["flower","flow","flight"]`
- **Output:** `"fl"`

<h3 style="color:#8250df;">Example 2</h3>

- **Input:** `strs = ["dog","racecar","car"]`
- **Output:** `""`
- **Explanation:** There is no common prefix among the input strings.

---

<h3 style="color:#1a7f37;">Constraints</h3>

- `1 <= strs.length <= 200`
- `0 <= strs[i].length <= 200`
- `strs[i]` consists of only lowercase English letters if it is non-empty.

```python
class Solution:
    def longestCommonPrefix(self, strs: list[str]) -> str:

        first = strs[0]
        
        for i in range(len(first)):
            ch = first[i]
            
            for word in strs[1:]:
                if i >= len(word) or word[i] != ch:
                    return first[:i]

        return first


sol = Solution()
print(sol.longestCommonPrefix(["flower", "flow", "flight"]))
print(sol.longestCommonPrefix(["dog", "racecar", "car"]))
```
