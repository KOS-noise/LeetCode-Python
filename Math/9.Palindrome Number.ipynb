<h2 style="color:#0969da;">9. Palindrome Number</h2>

[LeetCode 문제 링크](https://leetcode.com/problems/palindrome-number/)

Given an integer `x`, return `true` if `x` is a palindrome, and `false` otherwise.

---

<h3 style="color:#8250df;">Example 1</h3>

- **Input:** `x = 121`
- **Output:** `true`
- **Explanation:** 121 reads as 121 from left to right and from right to left.

<h3 style="color:#8250df;">Example 2</h3>

- **Input:** `x = -121`
- **Output:** `false`
- **Explanation:** From left to right, it reads -121. From right to left, it becomes 121-. Therefore it is not a palindrome.

<h3 style="color:#8250df;">Example 3</h3>

- **Input:** `x = 10`
- **Output:** `false`
- **Explanation:** Reads 01 from right to left. Therefore it is not a palindrome.

---

<h3 style="color:#1a7f37;">Constraints</h3>

- `-2^31 <= x <= 2^31 - 1`

> Follow up: Could you solve it without converting the integer to a string?

```python
class Solution:
    def isPalindrome(self, x: int) -> bool:
        s = str(x)
        length_x = len(s)
        print(length_x)

        for i in range(length_x // 2):
            if s[i] != s[length_x -1 -i]:
                return False
         
        return True


sol = Solution()
print(sol.isPalindrome(121))
print(sol.isPalindrome(-121))
print(sol.isPalindrome(10))
```
